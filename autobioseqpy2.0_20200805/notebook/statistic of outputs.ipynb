```python
import os, sys
```

```python
print(os.listdir('./'))
```

```python
outDict = {}

for f in os.listdir('./'):
    outFile = '%s/performance' %f
    if os.path.exists(outFile):
        feas = f.split('__')
        numFeas = len(feas)
        outDict.setdefault(numFeas,{})
        outDict[numFeas][f] = {}
        with open(outFile) as FID:
            s = False
            for l in FID:
                if 'Predicting Performance' in l:
                    s = True
                    continue
                if not ':' in l:
                    continue
                k,v = l.strip().split(':')
                outDict[numFeas][f][k] = v
        
```

```python
c = 0
for numFeas in outDict:
    print(numFeas,len(outDict[numFeas]))
    c += len(outDict[numFeas])
    
print('total',c)
```

看上去没啥问题，那就生成一个csv

```python
header = ['taskName','numFea','ACC','F1','Recall','Pre','MCC']
outName = 'sata.csv'
sep = ','
with open(outName, 'w') as FIDO:
    FIDO.write(sep.join(header))
    FIDO.write('\n')
    for numFeas in outDict:
        for f in outDict[numFeas]:
            tmpList = [f,str(numFeas)]
            for k in header[2:]:
                tmpList.append(outDict[numFeas][f][k])
            FIDO.write(sep.join(tmpList))
            FIDO.write('\n')
```
